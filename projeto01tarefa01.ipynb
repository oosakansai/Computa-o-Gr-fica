```python
from amplpy import AMPL, ampl_notebook
ampl = ampl_notebook(
    modules = ["gurobi"],
    licence_uuid = "default",
    g = globals()
)
```

```python
%%ampl_eval
# Definição de Variáveis
# quantidade em kgs
var cevada >= 0;
var aveia >= 0;
var soja >= 0;
var milho >= 0;

# Função Objetivo
minimize custo: 30*cevada + 48*aveia + 44*soja + 56*milho;

# Restrições
subject to R1: cevada + aveia + soja + milho = 10000;
subject to R2: 0.069*cevada + 0.085*aveia + 0.09*soja + 0.271*milho = 1500;
subject to R3: 0.06*cevada + 0.11*aveia + 0.11*soja + 0.14*milho >= 800;
subject to R4: 1760*cevada + 1700*aveia + 1056*soja + 1400*milho >= 1100 * 10000; # min calorias
subject to R5: 1760*cevada + 1700*aveia + 1056*soja + 1400*milho <= 2250 * 10000; # max calorias
subject to R6: milho >= 2000;
subject to R7: soja <= 1200;
```

```python
%%ampl_eval
option solver gurobi;
solve;
```
